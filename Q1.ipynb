```python
from astropy.io import fits
import numpy as np
import matplotlib.pyplot as plt

path = "/g/data/qp00/sz4640/ASTR_computing/Assugnment2/astro_comp_assignment02_2026/data/Q1.2.fits"

with fits.open(path) as hdul:
    data = hdul[1].data

mask = data["parallax"] > 0
absolute_g = data["phot_g_mean_mag"][mask] + 5 * np.log10(data["parallax"][mask]) - 10

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].scatter(absolute_g, data["bp_rp"][mask], s=2)
ax[0].set_ylabel("BP - RP")
ax[0].set_xlabel("Absolute G magnitude")
ax[0].invert_yaxis()
ax[0].set_title("Gaia BP - RP vs absolute G magnitude")

ax[1].scatter( data["h_m"], data["j_m"] - data["ks_m"], s=2)
ax[1].set_ylabel("J - Ks")
ax[1].set_xlabel("Apparent H magnitude")
ax[1].invert_yaxis()
ax[1].set_title("2MASS J-Ks vs apparent H magnitude")

plt.tight_layout()
plt.savefig("/g/data/qp00/sz4640/ASTR_computing/Assugnment2/astro_comp_assignment02_2026/figures/Q1.3.png", dpi=300)
plt.show()
```
